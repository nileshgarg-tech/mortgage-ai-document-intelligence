## Text CLeaning For NLP
In this notebook, I will practice some very common techniques to process text for NLP advancement.
This is very much a pre cursor to any NLP model training task on a new dataset.

```python
# Loading the text
with open("/content/text_sample.txt") as f:
  txt = f.read()
txt
```

```python
# Removing extra spaces and new lines
import re
txt = re.sub(r'\s+', ' ', txt).strip()
# replaces spaces, tabs and newlines with single space
txt
```

```python
# Converting_Text_To_Lowercase
txt = txt.lower()
txt
```

```python
## Removing Punctuation and Special Characters
import string
txt = txt.translate(str.maketrans('', '', string.punctuation))
txt

# --- Alternative for Removing Punctuation and Special Characters using Regex ---
# To remove punctuation using regex:
# import re
# import string
# txt_after_regex_punct_removal = re.sub(r'[' + re.escape(string.punctuation) + ']', '', txt)
# print("Text after punctuation removal (regex alternative):", txt_after_regex_punct_removal)
#
# This is an alternative to the `str.translate` method used in a previous cell (PE_mGPzNygnS).
# It's commented out here to maintain the current processing flow, but can be uncommented to test.
```

```python
## Fixing Encoding Issues
import unicodedata
import ftfy

txt  = ftfy.fix_text(txt)  # Fix corrupted text
print(txt)
txt = unicodedata.normalize('NFKD', txt).encode('ascii', 'ignore').decode()
print(txt)
```

```python
# Fixing Contractions
import contractions

text = "I'll be there, but I don't know if he's coming."
txt = contractions.fix(txt)
txt

```

```python
import nltk
from nltk.corpus import stopwords

# Download stopwords if not already downloaded
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    words = text.split()
    filtered_words = [word for word in words if word.lower() not in stop_words]
    return " ".join(filtered_words)

clean_text = remove_stopwords(txt)
print(clean_text)
```
