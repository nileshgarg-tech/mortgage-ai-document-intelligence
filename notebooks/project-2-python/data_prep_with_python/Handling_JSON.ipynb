# Handling Json in Python
In this notebook, We are practicing how to handle json in python

```python
## Importing Necessary Libraries
import json
import pandas as pd
```

```python
## Loading the JSON Data
with open('/content/sample.json', 'r') as f:
    json_data = json.load(f)

json_data
```

```python
type(json_data)
```

```python
orders_df = pd.json_normalize(json_data, record_path='orders')

# Display only the 'order_id' and 'total' columns
display(orders_df[['order_id', 'total']])
```

```python
flat_data_df = pd.json_normalize(
    json_data,
    record_path=['orders', 'items'],
    meta=['customer_id', ['orders', 'order_id'], ['orders', 'total']]
)

flat_data_df = flat_data_df.rename(columns={'orders.order_id': 'order_id', 'orders.total': 'total'})

display(flat_data_df[['customer_id', 'order_id', 'product', 'quantity', 'total']])
```

```python
flat_data_df.to_json("items.json", orient="records")
```
